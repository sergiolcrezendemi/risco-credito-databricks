```python
%sql
-- 0) Infraestrutura de hml (idempotente): schemas e volumes que o pipeline assume que existem
CREATE SCHEMA IF NOT EXISTS credito_hml.bronze;
CREATE SCHEMA IF NOT EXISTS credito_hml.silver;
CREATE SCHEMA IF NOT EXISTS credito_hml.gold;
CREATE SCHEMA IF NOT EXISTS credito_hml.ml;

CREATE VOLUME IF NOT EXISTS credito_hml.bronze.raw_landing;
CREATE VOLUME IF NOT EXISTS credito_hml.bronze.checkpoints;
CREATE VOLUME IF NOT EXISTS credito_hml.bronze.schemas;
CREATE VOLUME IF NOT EXISTS credito_hml.silver.checkpoints;
```

```python
%python
# 1) Copiar os CSVs de dev para hml
for ds in ["training", "scoring"]:
    dbutils.fs.cp(
        f"/Volumes/credito_dev/bronze/raw_landing/{ds}/",
        f"/Volumes/credito_hml/bronze/raw_landing/{ds}/",
        recurse=True,
    )

# 2) Conferir
display(dbutils.fs.ls("/Volumes/credito_hml/bronze/raw_landing/training/"))
display(dbutils.fs.ls("/Volumes/credito_hml/bronze/raw_landing/scoring/"))

# 3) Limpar checkpoint e schemaLocation de hml, por segurança
for p in [
    "/Volumes/credito_hml/bronze/checkpoints/give_me_some_credit_training_bronze/",
    "/Volumes/credito_hml/bronze/checkpoints/give_me_some_credit_scoring_bronze/",
]:
    dbutils.fs.rm(p, recurse=True)

display(dbutils.fs.ls("/Volumes/credito_hml/bronze/schemas/"))  # apague aqui as pastas de schema de training/scoring
```
