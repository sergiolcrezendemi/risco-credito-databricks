Rode sempre na máquina local, dentro do repositório git. Nem todos os três comandos têm o mesmo uso.   (instruções local e no databricks)

pre-commit install: roda uma única vez por clone do repositório. Ele instala o hook em .git/hooks, e depois disso o pre-commit roda sozinho a cada git commit. Só faz sentido onde existe .git, ou seja, na sua máquina.

pre-commit run --all-files: roda os hooks agora, em todos os arquivos. É o comando para validar antes de dar push. Também precisa de um repositório git.

pip install pre-commit: instala a ferramenta. Faça uma vez na máquina local (no seu projeto você pode usar uv add --dev pre-commit ou uv run pre-commit ...).

No Databricks não dá para validar do jeito que você tentou. A pasta /Workspace/Users/... não tem .git acessível no terminal web, então o pre-commit falha com not a git repository. Também não vale a pena tentar contornar isso lá.

Existem dois caminhos que funcionam no Databricks:

Git folder (Repos): se o projeto for clonado como Git folder, a interface permite commit e push, mas sem os hooks do pre-commit rodando. Você teria que conferir à mão detalhes como a quebra de linha final.
Rodar o ruff direto: ruff check . e ruff format --check . funcionam em qualquer pasta, sem git. Mas eles cobrem só uma parte do que o pre-commit faz. O end-of-file-fixer e o check yaml não são do ruff.

Fluxo recomendado

Edite e teste no Databricks, se preferir.
Antes do commit, na máquina local, rode pre-commit run --all-files.
Faça git add do que foi corrigido, commit e push.
O CI roda os mesmos hooks e passa, porque já estava tudo corrigido.

Como o CI usa o mesmo pre-commit run --all-files, o que passa localmente passa no GitHub Actions. Se o pre-commit install estiver ativo, isso acontece automaticamente no commit.
